```python
sales['Order Date'] = pd.to_datetime(sales['Order Date'])
sales['Delivery Date'] = pd.to_datetime(sales['Delivery Date'])

customers['Birthday'] = pd.to_datetime(customers['Birthday'])

stores['Open Date'] = pd.to_datetime(stores['Open Date'])
```

```python
sales_customers = pd.merge(sales, customers, on='CustomerKey', how='left')
```

```python
sales_customers_products = pd.merge(
    sales_customers,
    products,
    on='ProductKey',
    how='left'
)
```

```python
final_data = pd.merge(
    sales_customers_products,
    stores,
    on='StoreKey',
    how='left'
)
```

```python
final_data.head()
```

```python
final_data.shape
```

```python
final_data.describe
```

```python
#Question 1
# Which product categories generate the highest sales volume?
category_sales = final_data.groupby('Category')['Quantity'].sum().sort_values(ascending=False)
category_sales
```

```python
category_sales.plot(
    kind='bar',
    figsize=(8,5),
    title='Sales Volume by Category'
)

plt.xlabel('Category')
plt.ylabel('Quantity Sold')
plt.show()
```

```python
# Insight : The analysis shows that Computers generate the highest sales volume among all categories. The company should focus on maintaining inventory availability, optimizing pricing strategies, and promoting computer products to increase overall revenue.
```

```python
# Question 2
# Which brands contribute the highest sales volume?
brand_sales = final_data.groupby('Brand')['Quantity'].sum().sort_values(ascending=False)
brand_sales.head(10)
```

```python
brand_sales.head(10).plot(
    kind='bar',
    figsize=(10,5),
    title='Top 10 Brands by Sales Volume'
)

plt.xlabel('Brand')
plt.ylabel('Quantity Sold')
plt.show()
```

```python
# Insight : Contoso is the leading brand based on sales volume, indicating strong customer demand and brand preference. The company should prioritize stock management and marketing campaigns for high-performing brands like Contoso.
```

```python
# Question 3 
# Which products are the most popular among customers?
top_products = final_data.groupby('Product Name')['Quantity'].sum().sort_values(ascending=False)
top_products.head(10)
```

```python
top_products.head(10).plot(
    kind='barh',
    figsize=(10,6),
    title='Top 10 Products by Quantity Sold'
)

plt.xlabel('Quantity Sold')
plt.show()
```

```python
# Insight : WWI Desktop PC2.33 X2330 Black is the most popular product based on customer purchases. The business should ensure continuous availability of this product and analyze customer preferences to promote similar products.
```

```python
# Question 4
# Which countries have the highest customer demand?
country_orders = final_data['Country_x'].value_counts()
country_orders
```

```python
country_orders.plot(
    kind='bar',
    figsize=(8,5),
    title='Orders by Country'
)

plt.xlabel('Country')
plt.ylabel('Number of Orders')
plt.show()
```

```python
# Insight : The United States has the highest customer demand, making it the most important market for the business. The company should focus on customer retention, targeted promotions, and expansion strategies in this region.
```

```python
# Question 5
# Which cities have the highest number of customers?
city_customers = final_data['City'].value_counts().head(10)
city_customers
```

```python
city_customers.plot(
    kind='bar',
    figsize=(10,5),
    title='Top 10 Cities by Customers'
)

plt.xlabel('City')
plt.ylabel('Customers')
plt.show()
```

```python
# Insight : Toronto has the highest customer concentration, showing strong market potential. The company can improve customer engagement and run location-based marketing campaigns to increase sales in this city.
```

```python
# Question 6
#Which stores receive the highest number of orders?
store_orders = final_data['StoreKey'].value_counts()
store_orders.head(10)
```

```python
store_orders.head(10).plot(
    kind='bar',
    figsize=(8,5),
    title='Top Stores by Order Volume'
)

plt.xlabel('Store')
plt.ylabel('Orders')
plt.show()
```

```python
# Insight : Store 0 receives the highest number of orders compared to other stores. The company should analyze the performance factors of this store and apply successful strategies to improve other store locations.
```

```python
# Question 7
# What is the customer gender distribution?
gender_distribution = final_data['Gender'].value_counts()
gender_distribution
```

```python
gender_distribution.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6),
    title='Customer Distribution by Gender'
)

plt.ylabel('')
plt.show()
```

```python
# Insight : The customer base has a balanced gender distribution, with 31,804 male customers and 31,080 female customers. This indicates that products attract customers from both groups, allowing the company to create inclusive marketing strategies.
```

```python
# Question 8
# Which product subcategories are most popular?
subcategory_sales = final_data.groupby('Subcategory')['Quantity'].sum().sort_values(ascending=False)
subcategory_sales.head(10)
```

```python
subcategory_sales.head(10).plot(
    kind='barh',
    figsize=(10,6),
    title='Top Subcategories'
)

plt.xlabel('Quantity Sold')
plt.show()
```

```python
# Insight : Movie DVD are the most popular product subcategory, showing strong customer interest in entertainment products. The company should maintain sufficient stock and consider promotional offers to increase sales.
```

```python
# Question 9
# Are there seasonal trends in order volume?
monthly_orders = final_data.groupby(
    final_data['Order Date'].dt.month
)['Quantity'].sum()
monthly_orders
```

```python
monthly_orders.plot(
    kind='line',
    marker='o',
    figsize=(10,5),
    title='Monthly Order Trend'
)

plt.xlabel('Month')
plt.ylabel('Quantity Sold')
plt.show()
```

```python
# Insight : The analysis shows seasonal variation in sales, with the highest order volume occurring in December, followed by February and January. The company should prepare inventory, staffing, and promotional campaigns before peak sales periods.
```
